## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import numpy as np

df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

print("=" * 70)
print("REAL NUMBERS: Why Refresh Scoring is Worth 7 Weeks")
print("=" * 70)

print(f"\nDataset: {df.shape[0]:,} content items x {df.shape[1]} signals")

# NUMBER 1: Impressions vary widely
print("\n1. IMPRESSION SPREAD (90-day window)")
print(f"   Median: {df['impressions_90d'].median():,.0f}")
print(f"   Max: {df['impressions_90d'].max():,.0f}")
print(f"   Min: {df['impressions_90d'].min():,.0f}")
print(f"   >> 100x+ variance. Getting rank order RIGHT saves editor time.")

# NUMBER 2: Opportunity pool
print("\n2. OPPORTUNITY POOL (Low CTR + High Traffic)")
low_ctr = df['ctr'] < 0.75
high_traffic = df['impressions_90d'] > df['impressions_90d'].quantile(0.5)
pool = df[low_ctr & high_traffic]
print(f"   Pages with CTR <0.75% AND above-median impressions: {len(pool):,}")
print(f"   Total impressions: {pool['impressions_90d'].sum():,.0f}")
print(f"   If CTR +0.5ppt: ~{pool['impressions_90d'].sum() * 0.005:,.0f} more clicks")

# NUMBER 3: Signals available
print("\n3. RICH SIGNALS FOR SCORING")
for col in ['word_count', 'engagement_rate', 'ctr']:
    if col in df.columns:
        pct = df[col].notna().sum() / len(df) * 100
        print(f"   {col}: {pct:.1f}% available")

print("\n" + "=" * 70)
```
